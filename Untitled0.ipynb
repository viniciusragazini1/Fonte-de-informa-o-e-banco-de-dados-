```python

import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
import numpy as np

# 1. Enviar a planilha CSV
arquivo = "/content/student_exam_scores.csv"

# 2. Ler os dados
df = pd.read_csv(arquivo)

# Selecionar as variáveis da atividade
X = df[["hours_studied"]]
y = df["exam_score"]

# 3. Criar o modelo de regressão linear
modelo = LinearRegression()
modelo.fit(X, y)

# Calcular as previsões e métricas
previsoes = modelo.predict(X)
r2 = r2_score(y, previsoes)
correlacao = df["hours_studied"].corr(df["exam_score"])

# 4. Mostrar os resultados
print("RESULTADO DA REGRESSÃO LINEAR")
print(f"Intercepto: {modelo.intercept_:.4f}")
print(f"Inclinação: {modelo.coef_[0]:.4f}")
print(f"R²: {r2:.4f}")
print(f"Correlação: {correlacao:.4f}")

if modelo.coef_[0] > 0:
    print("\nConclusão: Sim, mais horas de estudo estão associadas a notas maiores.")
elif modelo.coef_[0] < 0:
    print("\nConclusão: Mais horas de estudo estão associadas a notas menores.")
else:
    print("\nConclusão: O modelo não indica uma associação linear.")

# 5. Criar o gráfico de dispersão e a reta de regressão
plt.figure(figsize=(8, 5))
plt.scatter(X["hours_studied"], y, alpha=0.7, label="Notas observadas")

x_linha = np.linspace(X["hours_studied"].min(),
                      X["hours_studied"].max(), 100)
y_linha = modelo.predict(pd.DataFrame({"hours_studied": x_linha}))

plt.plot(x_linha, y_linha, label="Regressão linear")
plt.xlabel("Horas de estudo")
plt.ylabel("Nota da prova")
plt.title("Horas de estudo x Nota da prova")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
